```python
import torch
import requests
import pandas as pd
```

```python
username = "switz2000"
headers={"User-Agent": "your-app/1.0 your@email.com"}

# Get most recent archive
url = f"https://api.chess.com/pub/player/{username}/games/archives"
archives = requests.get(url, headers=headers).json()
# print(archives)

latest_url = archives["archives"][-1]

N = 10

# Fetch games from latest month
games = requests.get(latest_url, headers=headers).json()["games"][-N:]

# Save each game's PGN
for i, game in enumerate(games):
    pgn = game.get("pgn", "")
    with open(f"game_{i+1}.pgn", "w") as f:
        f.write(pgn)

print(f"Saved {len(games)} games.")
```
